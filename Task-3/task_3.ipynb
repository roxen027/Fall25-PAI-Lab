```python
n = 4

val = [['.' for j in range(n)] for i in range(n)]
val[0][1] = 'Q'

for i in range(n):
    print('*' + '---+' * n)
    for j in range(n):
        print("|", val[i][j], end=' ')
    print("|")

print('*' + '---+' * n)
```

```python
n = 6

for i in range(n):
    for j in range(n):
        if i == 0 or i == n-1 or j == 0 or j == n-1:
            print("*", end=' ')
        else:
            print(" ", end=' ')
    print()
```

```python
n = 6

for i in range(n):
    for j in range(n):
        if (
            i == 0 or i == n-1 or      
            j == 0 or j == n-1 or      
            i == j or                 
            i + j == n - 1             
        ):
            print("*", end=' ')
        else:
            print(" ", end=' ')
    print()
```

```python
n = 6

for i in range(n):
    for j in range(n):
        if (
            i == 0 or i == n-1 or          # top & bottom border
            j == 0 or j == n-1 or          # left & right border
            (i == 2 and j in [1, 2, 4]) or # middle design
            (i == 3 and j in [1, 2, 4]) or
            (i == 4 and j == 3)
        ):
            print("*", end=' ')
        else:
            print(" ", end=' ')
    print()
```
